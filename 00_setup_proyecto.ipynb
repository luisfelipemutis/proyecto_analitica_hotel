### Objetivo
Preparar un entorno profesional para análisis de datos usando:

- Jupyter Notebook

- Estructura organizada
- Kernel independiente
- Gestión de dependencias

## Verificar versión de Python

```python
import sys

print("Versión de Python:")
print(sys.version)
```

### Definir nombre del proyecto

```python
from pathlib import Path

# Nombre del proyecto
proyecto = "proyecto_analitica_hotel"

# Ruta base
ruta_base = Path(proyecto)

print(f"Proyecto: {proyecto}")
```

# Crear estructura profesional del proyecto
## Primero crea una estructura organizada:

```python
carpetas = [
    "data/raw",
    "data/processed",
    "notebooks",
    "src",
    "models",
    "reports",
    "reports/figures"
]

for carpeta in carpetas:
    (ruta_base / carpeta).mkdir(parents=True, exist_ok=True)

print(" Estructura creada correctamente")
```

# Crear entorno virtual (venv)
Esta celda se ejecuta UNA sola vez.

```python
import os

ruta_venv = ruta_base / ".venv"

if not ruta_venv.exists():
    os.system(f"python -m venv {ruta_venv}")
    print("Entorno virtual creado")
else:
    print("El entorno virtual ya existe")
```

### Verificar estructura creada

```python
for ruta in ruta_base.rglob("*"):
    print(ruta)
```

# Activar el entorno virtual
para Windows

```python
import platform

sistema = platform.system()

print("Activar el entorno virtual desde la terminal:\n")

if sistema == "Windows":
    print(fr"{proyecto}\.venv\Scripts\activate")
else:
    print(f"source {proyecto}/.venv/bin/activate")
```

Luego ejecutas ese comando en la terminal.

## Crear archivo `.gitignore`

```python
contenido = """
.venv/
__pycache__/
.ipynb_checkpoints/
data/raw/
"""

with open(ruta_base / ".gitignore", "w") as archivo:
    archivo.write(contenido)

print(".gitignore creado")
```

## Crear archivo README.md

```python
readme = f"""
# {proyecto}

Proyecto de análisis de datos desarrollado con Python y Jupyter Notebook.

## Estructura

- data/: datos del proyecto
- notebooks/: notebooks de análisis
- src/: código reutilizable
- models/: modelos entrenados
- reports/: reportes y visualizaciones
"""

with open(ruta_base / "README.md", "w", encoding="utf-8") as archivo:
    archivo.write(readme)

print("README.md creado")
```
